```python
import pandas as pd
import numpy as np
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report, confusion_matrix
import warnings
from typing import Tuple, List, Dict
import logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

class FraudDetectionSystem:
    """
    Optimized system for credit card fraud detection using multiple methods
    """
    def __init__(self,
                 zscore_threshold: float = 3.5,
                 contamination: float = 0.1,
                 random_state: int = 42):
        self.zscore_threshold = zscore_threshold
        self.contamination = contamination
        self.random_state = random_state
        self.scaler = StandardScaler()
        self.isolation_forest = IsolationForest(
            contamination=contamination,
            random_state=random_state,
            n_jobs=-1
        )

    def load_and_preprocess_data(self, filepath: str) -> Tuple[pd.DataFrame, pd.DataFrame]:
        dtypes = {
            'Time': 'float32',
            'Amount': 'float32',
            'Class': 'int8'
        }
        chunks = []
        for chunk in pd.read_csv(filepath, dtype=dtypes, chunksize=10000):
            chunks.append(chunk)
        df = pd.concat(chunks, ignore_index=True)

        logger.info(f"Data loaded: {df.shape[0]} transactions")
        X = df.drop('Class', axis=1)
        y = df['Class']
        return X, y

    def detect_statistical_outliers(self, data: pd.DataFrame, columns: List[str]) -> np.ndarray:
        outlier_mask = np.zeros(len(data), dtype=bool)
        for column in columns:
            median = data[column].median()
            mad = stats.median_abs_deviation(data[column], nan_policy='omit')
            modified_z_scores = 0.6745 * (data[column] - median) / mad
            outlier_mask |= np.abs(modified_z_scores) > self.zscore_threshold

        return outlier_mask

    def prepare_features(self, X: pd.DataFrame) -> pd.DataFrame:
        X = X.copy()
        X['Amount_Log'] = np.log1p(X['Amount'])
        X['Time_Hour'] = X['Time'] % 24
        scaled_features = self.scaler.fit_transform(X)
        X_scaled = pd.DataFrame(scaled_features, columns=X.columns)

        return X_scaled

    def train_test_split_stratified(self,
                                  X: pd.DataFrame,
                                  y: pd.Series,
                                  test_size: float = 0.2) -> Tuple:
        return train_test_split(
            X, y,
            test_size=test_size,
            random_state=self.random_state,
            stratify=y
        )

    def detect_outliers_ensemble(self,
                               X: pd.DataFrame,
                               statistical_columns: List[str]) -> np.ndarray:

        statistical_outliers = self.detect_statistical_outliers(X, statistical_columns)
        isolation_forest_outliers = self.isolation_forest.fit_predict(X) == -1
        combined_outliers = statistical_outliers | isolation_forest_outliers

        return combined_outliers

    def process_and_train(self, filepath: str) -> Dict:
        try:
            X, y = self.load_and_preprocess_data(filepath)
            X_prepared = self.prepare_features(X)
            X_train, X_test, y_train, y_test = self.train_test_split_stratified(X_prepared, y)
            statistical_columns = ['Amount', 'Amount_Log', 'Time_Hour']
            train_outliers = self.detect_outliers_ensemble(X_train, statistical_columns)
            X_train_clean = X_train[~train_outliers]
            y_train_clean = y_train[~train_outliers]
            logger.info(f"Removed {train_outliers.sum()} outliers from training set")
            logger.info(f"Final training set shape: {X_train_clean.shape}")

            return {
                'X_train_clean': X_train_clean,
                'y_train_clean': y_train_clean,
                'X_test': X_test,
                'y_test': y_test,
                'outliers_removed': train_outliers.sum()
            }

        except Exception as e:
            logger.error(f"Error in processing pipeline: {str(e)}")
            raise

    def get_memory_usage(self, df: pd.DataFrame) -> str:
        memory_usage = df.memory_usage(deep=True).sum() / 1024**2
        return f"{memory_usage:.2f} MB"
if __name__ == "__main__":
    warnings.filterwarnings('ignore')
    fraud_detection = FraudDetectionSystem()
    results = fraud_detection.process_and_train("/content/drive/MyDrive/creditcard.csv")
    logger.info(f"Memory usage of clean training set: "
                f"{fraud_detection.get_memory_usage(results['X_train_clean'])}")
```

```python
import pandas as pd
import numpy as np
from typing import List, Dict, Optional
from dataclasses import dataclass
from datetime import datetime
import logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

@dataclass
class FeatureConfig:
    time_windows: List[str] = None
    high_value_percentile: float = 0.95
    unusual_hours: tuple = (5, 22)

    def __post_init__(self):
        if self.time_windows is None:
            self.time_windows = ['H', 'D', 'W']

class TransactionFeatureEngineer:

    def __init__(self, config: Optional[FeatureConfig] = None):
        self.config = config or FeatureConfig()
        self._feature_cache = {}

    def _cache_feature(self, name: str, data: pd.Series) -> None:
        self._feature_cache[name] = data

    def create_temporal_features(self, df: pd.DataFrame) -> pd.DataFrame:
        if not pd.api.types.is_datetime64_any_dtype(df['Time']):
            df['Time'] = pd.to_datetime(df['Time'], unit='s')

        features = {}
        features['hour_of_day'] = df['Time'].dt.hour
        features['day_of_week'] = df['Time'].dt.dayofweek
        features['is_weekend'] = (features['day_of_week'] >= 5).astype(int)
        hour = features['hour_of_day']
        features['unusual_time'] = ((hour < self.config.unusual_hours[0]) |
                                    (hour > self.config.unusual_hours[1])).astype(int)
        return pd.DataFrame(features)

    def create_amount_features(self, df: pd.DataFrame) -> pd.DataFrame:
        features = {}
        amount_stats = df.groupby('Class')['Amount'].agg(['mean', 'std'])
        features['avg_transaction_amount'] = df.groupby('Class')['Amount'].transform('mean')
        features['std_transaction_amount'] = df.groupby('Class')['Amount'].transform('std')
        features['deviation_from_avg'] = df['Amount'] - features['avg_transaction_amount']
        features['amount_to_avg_ratio'] = df['Amount'] / features['avg_transaction_amount']
        threshold = df['Amount'].quantile(self.config.high_value_percentile)
        features['high_value_transaction'] = (df['Amount'] > threshold).astype(int)
        return pd.DataFrame(features)

    def create_merchant_features(self, df: pd.DataFrame) -> pd.DataFrame:
        features = {}
        features['new_merchant'] = (df.groupby(df.index)['Class']
                                    .transform(lambda x: (x != x.shift()).cumsum())
                                    .fillna(1)
                                    .astype(int))
        time_window = '1h'
        transaction_counts = (df.groupby([df.index,
                                        pd.Grouper(key='Time', freq=time_window)])['Class']
                            .transform('count'))
        features['multiple_transactions'] = (transaction_counts > 1).astype(int)
        return pd.DataFrame(features)

    def engineer_features(self, df: pd.DataFrame, include_original: bool = True) -> pd.DataFrame:
        try:
            logger.info("Starting feature engineering pipeline...")
            df_copy = df.copy()
            feature_sets = {
                'temporal': self.create_temporal_features(df_copy),
                'amount': self.create_amount_features(df_copy),
                'merchant': self.create_merchant_features(df_copy)
            }
            engineered_features = pd.concat(feature_sets.values(), axis=1)
            if include_original:
                original_features = df_copy.drop(columns=['Class', 'Time'])
                final_features = pd.concat([engineered_features, original_features], axis=1)
            else:
                final_features = engineered_features
            logger.info(f"Feature engineering complete. Created {len(final_features.columns)} features")
            return final_features
        except Exception as e:
            logger.error(f"Error in feature engineering pipeline: {str(e)}")
            raise

    def get_feature_importance(self, df: pd.DataFrame, target: pd.Series) -> Dict[str, float]:
        features = self.engineer_features(df)
        importance = {column: abs(features[column].corr(target)) for column in features.columns}
        return dict(sorted(importance.items(), key=lambda x: x[1], reverse=True))
if __name__ == "__main__":
    config = FeatureConfig(
        time_windows=['H', 'D', 'W'],
        high_value_percentile=0.95,
        unusual_hours=(5, 22)
    )
    feature_engineer = TransactionFeatureEngineer(config)
    df = pd.read_csv("/content/drive/MyDrive/creditcard.csv")
    features = feature_engineer.engineer_features(df)
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

def filter_data_real_time(incoming_data, fraud_patterns, historical_data):
    required_columns = ['Time', 'Amount']
    if not all(column in incoming_data.columns for column in required_columns):
        raise ValueError(f"Input data must contain the following columns: {required_columns}")

    if not all(column in historical_data.columns for column in ['Amount']):
        raise ValueError("Historical data must contain the 'Amount' column.")
    try:
        incoming_data['transaction_hour'] = pd.to_datetime(incoming_data['Time'], unit='s').dt.hour
    except Exception as e:
        raise ValueError(f"Error parsing 'Time' column: {e}")
    incoming_data['fraud_flag'] = 0
    incoming_data.loc[(incoming_data['Amount'] > 1000) & (incoming_data['transaction_hour'] < 5), 'fraud_flag'] = 1
    if 'Location' in incoming_data.columns and 'high_risk_locations' in fraud_patterns:
        high_risk_locations = fraud_patterns['high_risk_locations']
        incoming_data.loc[incoming_data['Location'].isin(high_risk_locations), 'fraud_flag'] = 1
    if 'User_ID' in incoming_data.columns:
        incoming_data['transaction_count'] = incoming_data.groupby('User_ID')['User_ID'].transform('count')
        incoming_data.loc[incoming_data['transaction_count'] > 10, 'fraud_flag'] = 1
    try:
        historical_data['transaction_hour'] = pd.to_datetime(historical_data['Time'], unit='s').dt.hour
        features = ['Amount', 'transaction_hour']
        X_train = historical_data[features]
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        model = IsolationForest(contamination=0.01, random_state=42)
        model.fit(X_train_scaled)
        X_test = incoming_data[features]
        X_test_scaled = scaler.transform(X_test)
        incoming_data['anomaly_score'] = model.decision_function(X_test_scaled)
        incoming_data.loc[incoming_data['anomaly_score'] < -0.5, 'fraud_flag'] = 1
    except Exception as e:
        raise ValueError(f"Error during anomaly detection: {e}")
    filtered_data = incoming_data[incoming_data['fraud_flag'] == 1]

    return filtered_data
```

```python

from multiprocessing import Pool, cpu_count

def process_chunk(chunk, historical_mean, historical_std):
    chunk['transaction_hour'] = pd.to_datetime(chunk['Time'], unit='s').dt.hour
    chunk['fraud_flag'] = 0
    chunk.loc[(chunk['Amount'] > 1000) & (chunk['transaction_hour'] < 5), 'fraud_flag'] = 1
    chunk['amount_score'] = (chunk['Amount'] - historical_mean) / historical_std
    chunk.loc[chunk['amount_score'] > 3, 'fraud_flag'] = 1
    return chunk[chunk['fraud_flag'] == 1]

def filter_data_real_time_optimized(incoming_data, fraud_patterns, historical_data):
    required_columns = ['Time', 'Amount']
    if not all(column in incoming_data.columns for column in required_columns):
        raise ValueError(f"Input data must contain the following columns: {required_columns}")

    if not all(column in historical_data.columns for column in ['Amount']):
        raise ValueError("Historical data must contain the 'Amount' column.")
    historical_mean = historical_data['Amount'].mean()
    historical_std = historical_data['Amount'].std()
    num_cores = cpu_count()
    data_chunks = np.array_split(incoming_data, num_cores)
    with Pool(num_cores) as pool:
        results = pool.starmap(process_chunk, [(chunk, historical_mean, historical_std) for chunk in data_chunks])
    filtered_data = pd.concat(results)
    return filtered_data
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from imblearn.over_sampling import SMOTE

df = pd.read_csv("/content/drive/MyDrive/creditcard.csv")
X = df.drop(columns=['Class'])
y = df['Class']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

smote = SMOTE(sampling_strategy=0.5)
X_res, y_res = smote.fit_resample(X_scaled, y)

X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=42)

param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['auto', 'sqrt']
}

rf = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(estimator=rf, param_grid=param_grid,
                          cv=5, n_jobs=-1, verbose=2, scoring='recall')
grid_search.fit(X_train, y_train)

best_params = grid_search.best_params_
print(f"Best parameters: {best_params}")

best_rf = grid_search.best_estimator_
best_rf.fit(X_train, y_train)

y_pred = best_rf.predict(X_test)

print("\nOptimized Model Performance:")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
!pip install imblearn
import pandas as pd
from xgboost import XGBClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
from imblearn.over_sampling import SMOTE

RANDOM_STATE = 42

df = pd.read_csv("/content/drive/MyDrive/creditcard.csv")
X = df.drop(columns=["Class"])
y = df["Class"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

smote = SMOTE(sampling_strategy=0.5)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled, test_size=0.2, random_state=RANDOM_STATE)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    n_jobs=-1,
    random_state=RANDOM_STATE,
    early_stopping_rounds=10
)
xgb_model.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=False)
y_pred_xgb = xgb_model.predict(X_test)
print("XGBoost Results:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_xgb):.4f}")
print(classification_report(y_test, y_pred_xgb))

svm_model = SVC(
    kernel='rbf',
    C=1.0,
    gamma='scale',
    random_state=RANDOM_STATE,
    probability=True
)
svm_model.fit(X_train_scaled, y_train)
y_pred_svm = svm_model.predict(X_test_scaled)
print("\nSVM Results:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_svm):.4f}")
print(classification_report(y_test, y_pred_svm))

nn_model = MLPClassifier(
    hidden_layer_sizes=(100, 50),
    activation='relu',
    solver='adam',
    alpha=0.0001,
    batch_size=32,
    early_stopping=True,
    random_state=RANDOM_STATE,
    max_iter=200
)
nn_model.fit(X_train_scaled, y_train)
y_pred_nn = nn_model.predict(X_test_scaled)
print("\nNeural Network Results:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_nn):.4f}")
print(classification_report(y_test, y_pred_nn))
```

```python
import pandas as pd
from collections import deque
import random

class FraudInvestigationUnit:
    def __init__(self, automated_system_alerts):
        self.alerts = deque(automated_system_alerts)
        self.investigators = {
            "Alice": [],
            "Bob": [],
            "Charlie": []
        }
        self.cases = {}

    def assign_case(self, alert):
        """Assigns an alert to an investigator with the least workload."""
        investigator = min(self.investigators, key=lambda x: len(self.investigators[x]))

        self.investigators[investigator].append(alert)
        case_id = len(self.cases) + 1
        self.cases[case_id] = {"alert": alert, "investigator": investigator, "status": "Open"}
        print(f"Case {case_id} assigned to {investigator}. Alert: {alert}")

    def resolve_case(self, case_id, outcome):
        """Updates the status of a case."""
        if case_id in self.cases:
            self.cases[case_id]["status"] = outcome
            print(f"Case {case_id} resolved with outcome: {outcome}")
        else:
            print(f"Case {case_id} not found.")

    def process_alerts(self):
        """Processes alerts in the queue."""
        while self.alerts:
            alert = self.alerts.popleft()
            self.assign_case(alert)
```

```python

alerts = ["Suspicious transaction from IP: 123.456.789.0",
          "High-value transaction at unusual time",
          "Multiple failed login attempts"]
investigation_unit = FraudInvestigationUnit(alerts)

investigation_unit.process_alerts()

investigation_unit.resolve_case(1, "Fraudulent")

```
